```python
import os
import pandas as pd

# For visual checks (optional)
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set the relative path to your data folder
DATA_DIR = '../NBAdata'

# Load CSV files
games = pd.read_csv(os.path.join(DATA_DIR, 'Games.csv'))
schedule = pd.read_csv(os.path.join(DATA_DIR, 'LeagueSchedule24_25.csv'))
players = pd.read_csv(os.path.join(DATA_DIR, 'Players.csv'))
player_stats = pd.read_csv(os.path.join(DATA_DIR, 'PlayerStatistics.csv'))
team_histories = pd.read_csv(os.path.join(DATA_DIR, 'TeamHistories.csv'))
team_stats = pd.read_csv(os.path.join(DATA_DIR, 'TeamStatistics.csv'))
```

```python
def explore_dataset(name, df):
    print(f"=== {name} ===")
    print(f"Shape: {df.shape}")
    print("Columns:", df.columns.tolist())
    print("Missing Values:\n", df.isnull().sum())
    display(df.head(3))
    print("\n" + "="*50 + "\n")
```

```python
datasets = {
    'Games': games,
    'League Schedule': schedule,
    'Players': players,
    'Player Statistics': player_stats,
    'Team Histories': team_histories,
    'Team Statistics': team_stats
}

for name, df in datasets.items():
    explore_dataset(name, df)
```

```python
plt.figure(figsize=(10, 4))
sns.histplot(games['homeScore'], color='skyblue', label='Home', bins=30)
sns.histplot(games['awayScore'], color='salmon', label='Away', bins=30)
plt.title('Final Score Distribution')
plt.xlabel('Score')
plt.ylabel('Count')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
games['home_win'] = (games['winner'] == games['hometeamId']).astype(int)
home_win_rate = games['home_win'].mean()

print(f"Overall Home Win Rate: {home_win_rate:.2%}")
```

```python
top_scorers = player_stats.groupby(['personId', 'firstName', 'lastName'])['points'] \
    .sum().sort_values(ascending=False).head(10)

top_scorers.plot(kind='bar', color='purple', figsize=(10, 4))
plt.title('Top 10 Scorers (Total Points)')
plt.ylabel('Points')
plt.xticks(
    ticks=range(len(top_scorers)),
    labels=[f"{fn} {ln}" for _, fn, ln in top_scorers.index],
    rotation=45,
    ha='right'
)
plt.tight_layout()
plt.show()

```

```python
top_assists = player_stats.groupby(['personId', 'firstName', 'lastName'])['assists'] \
    .sum().sort_values(ascending=False).head(10)

top_assists.plot(kind='bar', color='orange', figsize=(10, 4))
plt.title('Top 10 Assist Leaders')
plt.ylabel('Assists')
plt.xticks(
    ticks=range(len(top_assists)),
    labels=[f"{fn} {ln}" for _, fn, ln in top_assists.index],
    rotation=45,
    ha='right'
)
plt.tight_layout()
plt.show()
```

```python
top_rebounds = player_stats.groupby(['personId', 'firstName', 'lastName'])['reboundsTotal'] \
    .sum().sort_values(ascending=False).head(10)

top_rebounds.plot(kind='bar', color='teal', figsize=(10, 4))
plt.title('Top 10 Rebounders')
plt.ylabel('Rebounds')
plt.xticks(
    ticks=range(len(top_rebounds)),
    labels=[f"{fn} {ln}" for _, fn, ln in top_rebounds.index],
    rotation=45,
    ha='right'
)
plt.tight_layout()
plt.show()
```
